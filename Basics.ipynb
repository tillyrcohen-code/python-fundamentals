```python
# 1. Creating a list of London stations
stations = ["Upminster", "Barking", "West Ham", "Fenchurch Street"]

# 2. Let's see the whole list
print("Our complete route:")
print(stations)

# 2.1. Clean up list for user - glue items together using .join()
clean_stations = ", ".join(stations)
print("\nOur complete route (clean version):")
print(clean_stations)

# 2.2. * Method - tells Python to unpack the box
print(*stations)

# 3. Accessing specific items using their index (position)
print(f"\nThe starting station is: {stations[0]}")
print(f"The terminal station is: {stations[3]}")

# 4. Adding a new item to the end of the list using .append()
stations.append("Limehouse")
print(f"\nUpdated route with new stop: {stations}")

# 5. Test - print just Barking
print(f"\nThe second station is: {stations[1]}")
```

```python
# Loop Practice
stations = ["Upminster", "Barking", "West Ham", "Fenchurch Street"]

print("Starting Automated Route Announcements:\n")

# This loop grabs each station one by one and calls it 'stop'
for stop in stations:
  # Everything indented here happens to EACH  station
  print(f"The next station is {stop}. Change here for connecting services.")
  print("---")

# This line is NOT indented, so it only runs once at the very end
print("End of the line. Please get off.")
```

```python
# Conditionals Practice

stations = ["Upminster", "Barking", "West Ham", "Fenchurch Street"]

for stop in stations:
  print(f"Arriving at: {stop}")

  # 1. Check if the station is Barking
  if stop == "Barking":
    print("Change here for the District Line and Overground.")

  # 2. If it wasn't Barking, check if it's West Ham
  elif stop == "West Ham":
    print("Change here for the Jubilee, District, and DLR lines.")

  # 2.2 Is it Fenchurch Street
  elif stop == "Fenchurch Street":
    print("Terminus station. Please get off here.")

  # 3. If it wasn't either of those, do this by default
  else:
    print("This is a regular stop.")

  print("-" * 30)  # Prints a neat divider line
```

```python
# Dictionaries Practice - used to group related info. together

# 1. Creating a single dictionary representing one station
station_profile =  {
    "name": "Upminster",
    "zone": 6,
    "ticket_price": 16.50,
    "is_underground": False
}

# 2. To get a specific piece of data out, you look it up by its key
print("--- Single Station Profile ---")
print(f"Station Name: {station_profile['name']}")  # Have to use '' for name as used "" already for the f string
print(f"Fare Zone: {station_profile['zone']}")

# 3. You can easily update data inside a dictionary too
station_profile["is_underground"] = True # Maybe the tracks changed!
print(f"Updated Underground Status: {station_profile['is_underground']}")
```

```python
# Combine Lists, Dictionaries, For Loops, and Conditionals

# A list containing multiple station dictionaries (mimicking real API data)
live_network_data = [
    {
        "name": "Upminster",
        "zone": 6,
        "ticket_price": 16.50,
        "is_underground": False
    },
    {
        "name": "Barking",
        "zone": 4,
        "ticket_price": 5.30,
        "is_underground": True
    },
    {
        "name": "West Ham",
        "zone": 3,
        "ticket_price": 4.80,
        "is_underground": True
    }
]

print("Scanning London Transit Network Data...\n")

# Loop through each dictionary inside the list
for station in live_network_data:
  print(f"Checking Station: {station['name']} (Zone {station['zone']})")

  # Check if a ticket costs more than £10.00
  if station["ticket_price"] > 10.00:
    print(f" ALERT: High fare warning! Price is £{station['ticket_price']:.2f}")

  # Check if it's part of the Underground
  if station["is_underground"] == True:
    print("Connection: This is an Underground Tube Station.")
  else:
    print("Connection: This is a National Rail / Overground station.")

    print("-" * 40)
```

```python
# Functons Practice
# 1. Defining the function using def
def announce_welcome():
  print("Welcome to the Transport for London Network Monitor.")
  print("System Status: Operational")
  print("-------------------------------------")

# 2. Calling the function
announce_welcome()
announce_welcome()

# 3. We define the function and tell it to expect a list of data, which we'll call 'network_list'
def scan_stations(network_list):
  print("Scanning Network Data via Function...\n")

  for station in network_list:
    print(f"Checking Station {station['name']}")

    if station["ticket_price"] > 10.00:
      print(f"ALERT: High Fare Warning! (£{station['ticket_price']:.2f})")

    if station["is_underground"]:
      print("Underground Connection Available.")
    else:
      print("National Rail Connection Available.")
    print("-" * 30)

# 4. This is our raw data
my_stations = [
    {"name": "Upminster", "ticket_price": 16.50, "is_underground": False},
    {"name": "Barking", "ticket_price": 5.30, "is_underground": True}
]

# 5. Now we run it...
scan_stations(my_stations)
```
